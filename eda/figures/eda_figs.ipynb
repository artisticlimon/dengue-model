```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def filtrar_provincias(df, prov):
    columns = [df.columns.to_list().index(l) for l in df.columns if l.startswith(prov)]
    provincia = df.iloc[:, columns]
    weeks = df["week"]
    provincia = pd.concat([weeks, provincia], axis = 1)

    return(provincia)

def ticks_years(x, y, n, ax):
    [l.set_visible(False) for (i,l) in enumerate(ax[x, y].xaxis.get_ticklabels()) if i % n != 0]
    for i, tick in enumerate(ax[x, y].xaxis.get_major_ticks()):
        if i % n != 0:
            tick.tick1line.set_visible(False)
            tick.tick2line.set_visible(False)
            tick.gridline.set_visible(False) 

def boxplots_chir(df, var):
    sj = filtrar_provincias(df, "1")
    alajuela = filtrar_provincias(df, "2")
    cartago = filtrar_provincias(df, "3")
    heredia = filtrar_provincias(df, "4")
    guanacaste = filtrar_provincias(df, "5")
    puntarenas = filtrar_provincias(df, "6")
    limon = filtrar_provincias(df, "7")

    fig, ax = plt.subplots(4, 2, figsize=(20, 24))

    ax[0, 0].boxplot(sj.iloc[:, 1:], tick_labels = sj.columns[1:])
    ax[0, 0].set_title("San José")
    ax[0, 0].set_xlabel("Código de cantón")
    ax[0, 0].set_ylabel(var)

    ax[0, 1].boxplot(alajuela.iloc[:, 1:], tick_labels = alajuela.columns[1:])
    ax[0, 1].set_title("Alajuela")
    ax[0, 1].set_xlabel("Código de cantón")
    ax[0, 1].set_ylabel(var)

    ax[1, 0].boxplot(cartago.iloc[:, 1:], tick_labels = cartago.columns[1:])
    ax[1, 0].set_title("Cartago")
    ax[1, 0].set_xlabel("Código de cantón")
    ax[1, 0].set_ylabel(var)

    ax[1, 1].boxplot(heredia.iloc[:, 1:], tick_labels = heredia.columns[1:])
    ax[1, 1].set_title("Heredia")
    ax[1, 1].set_xlabel("Código de cantón")
    ax[1, 1].set_ylabel(var)

    ax[2, 0].boxplot(guanacaste.iloc[:, 1:], tick_labels = guanacaste.columns[1:])
    ax[2, 0].set_title("Guanacaste")
    ax[2, 0].set_xlabel("Código de cantón")
    ax[2, 0].set_ylabel(var)

    ax[2, 1].boxplot(puntarenas.iloc[:, 1:], tick_labels = puntarenas.columns[1:])
    ax[2, 1].set_title("Puntarenas")
    ax[2, 1].set_ylabel(var)
    ax[2, 1].set_xlabel("Código de cantón")

    ax[3, 0].boxplot(limon.iloc[:, 1:], tick_labels = limon.columns[1:])
    ax[3, 0].set_title("Limón")
    ax[3, 0].set_ylabel(var)
    ax[3, 0].set_xlabel("Código de cantón")
    ax[3, 0].set_position([0.3, 0.07, 0.4, 0.2])

    ax[3, 1].set_visible(False)

    plt.show()

def serie_chir(df, var, cantones):

    fig, ax = plt.subplots(4, 2, figsize=(20, 24))

    ax[0, 0].plot(df["week"], df[cantones[0]])
    ax[0, 0].set_title("San José")
    ax[0, 0].set_ylabel(var)
    ax[0, 0].set_xlabel("Semana")
    ticks_years(0, 0, 52, ax)

    ax[0, 1].plot(df["week"], df[cantones[1]])
    ax[0, 1].set_title("Alajuela")
    ax[0, 1].set_ylabel(var)
    ax[0, 1].set_xlabel("Semana")
    ticks_years(0, 1, 52, ax)

    ax[1, 0].plot(df["week"], df[cantones[2]])
    ax[1, 0].set_title("Cartago")
    ax[1, 0].set_ylabel(var)
    ax[1, 0].set_xlabel("Semana")
    ticks_years(1, 0, 52, ax)

    ax[1, 1].plot(df["week"], df[cantones[3]])
    ax[1, 1].set_title("Heredia")
    ax[1, 1].set_ylabel(var)
    ax[1, 1].set_xlabel("Semana")
    ticks_years(1, 1, 52, ax)

    ax[2, 0].plot(df["week"], df[cantones[4]])
    ax[2, 0].set_title("Guanacaste")
    ax[2, 0].set_ylabel(var)
    ax[2, 0].set_xlabel("Semana")
    ticks_years(2, 0, 52, ax)

    ax[2, 1].plot(df["week"], df[cantones[5]])
    ax[2, 1].set_title("Puntarenas")
    ax[2, 1].set_ylabel(var)
    ax[2, 1].set_xlabel("Semana")
    ticks_years(2, 1, 52, ax)

    ax[3, 0].plot(df["week"], df[cantones[6]])
    ax[3, 0].set_title("Limón")
    ax[3, 0].set_ylabel(var)
    ax[3, 0].set_xlabel("Semana")
    ax[3, 0].set_position([0.3, 0.07, 0.4, 0.2])
    ticks_years(3, 0, 52, ax)

    ax[3, 1].set_visible(False)

    plt.show()
```

```python
lst_mean = pd.read_csv("../../data/raw/lst_mean_eda.csv")
lst_mean.drop("Unnamed: 0", axis =1, inplace=True)

boxplots_chir(lst_mean, "Temperatura promedio")
```

```python
serie_chir(lst_mean, "Temperatura promedio", ["111", "212", "308", "401", "508", "608", "701"])
```

```python
lst_max = pd.read_csv("../../data/raw/lst_max_eda.csv")
lst_max.drop("Unnamed: 0", axis =1, inplace=True)

boxplots_chir(lst_max, "Temperatura máxima")

serie_chir(lst_max, "Temperatura máxima", ["111", "212", "308", "401", "508", "608", "701"])
```

```python
lst_min = pd.read_csv("../../data/raw/lst_min_eda.csv")
lst_min.drop("Unnamed: 0", axis =1, inplace=True)

boxplots_chir(lst_min, "Temperatura mímna")

serie_chir(lst_min, "Temperatura mínima", ["111", "212", "308", "401", "508", "608", "701"])
```

```python
lst_median = pd.read_csv("../../data/raw/lst_median_eda.csv")
lst_median.drop("Unnamed: 0", axis =1, inplace=True)

boxplots_chir(lst_median, "Mediana de la temperatura")

serie_chir(lst_median, "Mediana de la temperatura", ["111", "212", "308", "401", "508", "608", "701"])
```
