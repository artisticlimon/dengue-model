```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cases2020 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2020", header = 0)
cases2021 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2021", header = 0)
cases2022 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2022", header = 0)
cases2023 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2023", header = 0)
cases2024 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2024", header = 0)
cases2025 = pd.read_excel("../../data/raw/dengue_limpio.xlsx", sheet_name="2025", header = 0)
```

```python
cases2020 = cases2020.set_index("ID")
cases2021 = cases2021.set_index("ID")
cases2022 = cases2022.set_index("ID")
cases2023 = cases2023.set_index("ID")
cases2024 = cases2024.set_index("ID")
cases2025 = cases2025.set_index("ID")
```

```python
casos = pd.read_csv("../../data/raw/cases_dengue_rr.csv")
```

```python
casos.head()
```

```python
fig, ax = plt.subplots(4, 2, figsize=(10, 12))

def ticks_years(x, y, n):
    [l.set_visible(False) for (i,l) in enumerate(ax[x, y].xaxis.get_ticklabels()) if i % n != 0]
    for i, tick in enumerate(ax[x, y].xaxis.get_major_ticks()):
        if i % n != 0:
            tick.tick1line.set_visible(False)
            tick.tick2line.set_visible(False)
            tick.gridline.set_visible(False) 


ax[0, 0].plot(casos["semana"], casos["0"])
ax[0, 0].set_title("Total")
ax[0, 0].set_ylabel("Casos")
ax[0, 0].set_xlabel("Semana")
ticks_years(0, 0, 52)

ax[0, 1].plot(casos["semana"], casos["1"])
ax[0, 1].set_title("San José")
ax[0, 1].set_ylabel("Casos")
ax[0, 1].set_xlabel("Semana")
ticks_years(0, 1, 52)

ax[1, 0].plot(casos["semana"], casos["2"])
ax[1, 0].set_title("Alajuela")
ax[1, 0].set_ylabel("Casos")
ax[1, 0].set_xlabel("Semana")
ticks_years(1, 0, 52)

ax[1, 1].plot(casos["semana"], casos["3"])
ax[1, 1].set_title("Cartago")
ax[1, 1].set_ylabel("Casos")
ax[1, 1].set_xlabel("Semana")
ticks_years(1, 1, 52)

ax[2, 0].plot(casos["semana"], casos["4"])
ax[2, 0].set_title("Heredia")
ax[2, 0].set_ylabel("Casos")
ax[2, 0].set_xlabel("Semana")
ticks_years(2, 0, 52)

ax[2, 1].plot(casos["semana"], casos["5"])
ax[2, 1].set_title("Guanacaste")
ax[2, 1].set_ylabel("Casos")
ax[2, 1].set_xlabel("Semana")
ticks_years(2, 1, 52)

ax[3, 0].plot(casos["semana"], casos["6"])
ax[3, 0].set_title("Puntarenas")
ax[3, 0].set_ylabel("Casos")
ax[3, 0].set_xlabel("Semana")
ticks_years(3, 0, 52)

ax[3, 1].plot(casos["semana"], casos["7"])
ax[3, 1].set_title("Limón")
ax[3, 1].set_ylabel("Casos")
ax[3, 1].set_xlabel("Semana")
ticks_years(3, 1, 52)

fig.tight_layout(rect=[0.1, 0.05, 1, 0.95])
```
