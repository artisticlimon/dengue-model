# Calculating relative risk

```python
import pandas as pd
```

```python
cases = pd.read_csv("../data/raw/cases_dengue_rr.csv")
```

```python
cases.head()
```

```python
cases.shape
```

```python
import numpy as np
pob_dummy= pd.DataFrame(np.random.randint(low = 500, high = 900, size=(318, 93)), columns=cases.columns[0:93])
pob_dummy.head()
```

```python
relative_risk_num = cases.iloc[:, 1:94].div(pob_dummy.iloc[:, 1:94], axis="columns")
```

```python
relative_risk_num
```

```python
relative_risk_den = cases["0"].div(pob_dummy["0"])
```

```python
relative_risk_den.head()
```

```python
relative_risk = relative_risk_num.div(relative_risk_den.values, axis=0)
```

```python
relative_risk["semana"] = cases["semana"]
```

```python
relative_risk
```

```python
relative_risk = pd.melt(relative_risk , id_vars="semana", var_name = "canton", value_name="casos")
relative_risk ["week-canton"] = relative_risk ["semana"] + "-" + relative_risk ["canton"].astype(str)
relative_risk .drop(columns=["semana", "canton"], inplace=True)
```

```python
relative_risk.head()
```

```python
relative_risk.to_csv("../data/clean/relative_risk.csv", index=False)
```
