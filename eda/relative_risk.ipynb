# Calculating relative risk

```python
import pandas as pd
```

```python
cases = pd.read_csv("../data/clean/casos_dengue.csv")
```

```python
cases.head()
```

```python
cases.shape
```

```python
import numpy as np
pob_dummy= pd.DataFrame(np.random.randint(low = 500, high = 900, size=(318, 93)), columns=cases.columns[0:93])
pob_dummy.head()
```

```python
relative_risk_num = cases.iloc[:, 1:94].div(pob_dummy.iloc[:, 1:94], axis="columns")
```

```python
relative_risk_num
```

```python
relative_risk_den = cases["0"].div(pob_dummy["0"])
```

```python
relative_risk_den.head()
```

```python
relative_risk = relative_risk_num.div(relative_risk_den.values, axis=0)
```

```python
relative_risk
```
