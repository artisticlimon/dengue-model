# Calculating relative risk

```python
import pandas as pd
```

```python
cases = pd.read_csv("../data/raw/cases_dengue_rr.csv")
```

```python
cases.head()
```

```python
cases.shape
```

```python
cases.drop(columns=["1"], inplace=True)
```

```python
cases.head()
```

```python
pob = pd.read_csv("../data/raw/pob_proj_rr.csv")
```

```python
pob.head()
```

```python
relative_risk_num = cases.iloc[:, 0:93].div(pob.iloc[:, 0:93], axis="columns")
```

```python
relative_risk_num
```

```python
relative_risk_den = cases.div(pob.iloc[:, 1:93], axis="columns")
```

```python
relative_risk_den.head()
```

```python
relative_risk_den.drop(columns=["2", "3", "4", "5", "6", "7"], inplace=True)
```

```python
relative_risk = relative_risk_num.iloc[:, :relative_risk_den.shape[1]].div(relative_risk_den.values, axis=0)
```

```python
relative_risk.head()
```

```python
relative_risk["semana"] = cases["semana"]
```

```python
relative_risk
```

```python
relative_risk = pd.melt(relative_risk , id_vars="semana", var_name = "canton", value_name="casos")
relative_risk ["week-canton"] = relative_risk ["semana"] + "-" + relative_risk ["canton"].astype(str)
relative_risk .drop(columns=["semana", "canton"], inplace=True)
```

```python
relative_risk.head()
```

```python
relative_risk.to_csv("../data/clean/relative_risk.csv", index=False)
```
