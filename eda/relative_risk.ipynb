# Calculating relative risk

```python
import pandas as pd
import numpy as np
```

```python
cases = pd.read_csv("../data/raw/cases_dengue_rr.csv")
```

```python
cases.head()
```

```python
cases.shape
```

```python
cases.drop(columns=["1"], inplace=True)
```

```python
cases.head()
```

```python
cases.drop(columns=["2", "3", "4", "5", "6", "7", '99', '98'], inplace=True)
```

```python
cases.head()
```

```python
cases.shape
```

```python
pob = pd.read_csv("../data/raw/pob_proj_rr.csv")
```

```python
pob.head()
```

```python
pob.shape
```

```python
cases.tail()
```

```python
relative_risk_num = cases.div(pob, axis="columns")
```

```python
relative_risk_num.tail()
```

```python
relative_risk_den = cases["0"].div(pob["0"], axis="index")
```

```python
relative_risk_den.head()
```

```python
relative_risk = relative_risk_num.div(relative_risk_den, axis="index")
```

```python
relative_risk.tail()
```

```python
relative_risk.drop(columns=["0", "index"], inplace=True)
```

```python
relative_risk["semana"] = cases["semana"]
```

```python
relative_risk
```

```python
relative_risk = pd.melt(relative_risk , id_vars="semana", var_name = "canton", value_name="rr")
relative_risk ["week-canton"] = relative_risk ["semana"] + "-" + relative_risk ["canton"].astype(str)
relative_risk .drop(columns=["semana", "canton"], inplace=True)
```

```python
umbral_bajo = np.nanpercentile(relative_risk["rr"], q = 30)
umbral_medio = np.nanpercentile(relative_risk["rr"], q = 70)
umbral_alto = max(relative_risk["rr"])

relative_risk["clasi_rr"] = np.where(relative_risk["rr"] <= umbral_bajo, "Bajo", np.where(relative_risk["rr"] >= umbral_medio, "Alto", "Medio"))
```

```python
relative_risk.head()
```

```python
relative_risk.tail()
```

```python
relative_risk.to_csv("../data/clean/relative_risk.csv", index=False)
```
