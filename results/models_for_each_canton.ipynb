## Initial configuration

```python
import pandas as pd
import sys
import numpy as np

sys.path.append('../models')

from model_rr import Model
```

```python
df = pd.read_csv('../data/clean/full_data.csv')
```

```python
cantons = [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115, 116, 117, 118, 119, 120, 201, -202, 203, 204, 205, 206, 207, 208, 209, 210, 211, 212, 213, 214, 215, 216, 301, 302, 303, 304, 305, 306, 307, 308, 401, 402, 403, 404, 405, 406, 407, 408, 409, 410, 501, 502, 503, 504, 505, 506, 507, 508, 509, 510, 511, 601, 602, 603, 604, 605, 606, 607, 608, 609, 610, 611, 701, 702, 703, 704, 705, 706]

cantons = [str(x) for x in cantons]
```

# Random Forest

```python
cantons_rf = []
comp_time_rf_nofull = []

for canton in cantons: 
    instancia = Model(df, canton)
    X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = instancia.partition()
    elapsed_time = instancia.rf_reg(X_combined, y_combined, X_test, y_test, test, pds, epsilon)
    comp_time_rf_nofull.append(comp_time_rf_nofull)
    cantons_rf.append(canton)
    print(f"Model ready for {canton}")

comp_time_rf_nofull_df = pd.DataFrame({"canton": cantons_rf, "time": comp_time_rf_nofull})

comp_time_rf_nofull_df.to_csv('../data/model_results/comp_time/rf_comp_time_nofull.csv', index = False)
```

# XGB

```python
cantons_xgb = []
comp_time_xgb_nofull = []

for canton in cantons: 
    instancia = Model(df, canton)
    X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = instancia.partition()
    elapsed_time = instancia.xgb_reg(X_combined, y_combined, X_test, y_test, test, pds, epsilon)
    comp_time_xgb_nofull.append(comp_time_xgb_nofull)
    cantons_xgb.append(canton)

comp_time_xgb_nofull_df = pd.DataFrame({"canton": cantons_xgb, "time": comp_time_xgb_nofull})

comp_time_xgb_nofull_df.to_csv('../data/model_results/comp_time/xgb_comp_time_nofull.csv', index = False)
```

# HRF

## Fit models

```python
cantons_hrf = []
best_th_hrf = []
comp_time_hrf_nofull = []

for canton in cantons: 
    instancia = Model(df, canton)
    X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = instancia.partition()
    best_th_temp, elapsed_time = instancia.hybrid("hrf", X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon)
    comp_time_hrf_nofull.append(comp_time_hrf_nofull)
    print(best_th_temp)
    cantons_hrf.append(canton)
    best_th_hrf.append(best_th_temp)

comp_time_hrf_nofull_df = pd.DataFrame({"canton": cantons_hrf, "time": comp_time_hrf_nofull})

comp_time_hrf_nofull_df.to_csv('../data/model_results/comp_time/hrf_comp_time_nofull.csv')

best_th_hrf_df = pd.DataFrame({"canton": cantons_hrf, "best_th": best_th_hrf})

best_th_hrf_df.to_csv("../data/model_results/thresholds/thresholds_hrf.csv", index = False)
```
