# Comparing NRMSE for test

```python
import pandas as pd
import sys
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

project_root = Path.cwd().parent

sys.path.insert(0, str(project_root))

from model_rr import Model
```

```python
cantons = [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115, 116, 117, 118, 119, 120, 201, -202, 203, 204, 205, 206, 207, 208, 209, 210, 211, 212, 213, 214, 215, 216, 301, 302, 303, 304, 305, 306, 307, 308, 401, 402, 403, 404, 405, 406, 407, 408, 409, 410, 501, 502, 503, 504, 505, 506, 507, 508, 509, 510, 511, 601, 602, 603, 604, 605, 606, 607, 608, 609, 610, 611, 701, 702, 703, 704, 705, 706]


cantons = [str(x) for x in cantons]
```

# RF

```python
df = pd.read_csv('/home/marianne.pena/mainMarianne/dengue-model/data/full_data.csv')
```

```python
cod_cant_rf = []
point_rf = []
lower_rf = []
upper_rf = []
nrmse_rf = []

for canton in cantons: 
    instancia = Model(df) 
    X_combined, y_combined, X_test, X_test, y_test, test, pds, epsilon = instancia.partition()
    canton_temp, est_temp = instancia.calculate_test_nrmse("rf", epsilon)
    canton_temp, point_temp, lower_temp, upper_temp = instancia.confint_nrmse("rf", X_combined, y_combined, X_test, y_test, epsilon, which_set = "test", n_bootstraps = 500)
    cod_cant_rf.append(canton_temp)
    point_rf.append(point_temp)
    lower_rf.append(lower_temp)
    upper_rf.append(upper_temp)
    nrmse_rf.append(est_temp)
    print(f"Confint ready for {canton}")
```

```python
confint_rf = pd.DataFrame({"canton": cod_cant_rf,
                         "nrmse": nrmse_rf,
                         "mean": point_rf,
                         "lower": lower_rf, 
                         "upper": upper_rf})

confint_rf.to_csv("/home/marianne.pena/mainMarianne/dengue-model/data/confints/1rf_confint_test.csv", index = False)
```
