## Save metrics

```python
cod_cant_rf = []
nrmse_rf = []

for canton in cantons: 
    instancia = Model(df, canton)
    cant_temp, nrmse_temp = instancia.calculate_nrmse("rf")
    cod_cant_rf.append(canton)
    nrmse_rf.append(nrmse_temp)

nrmse_rf = pd.DataFrame({"canton": cod_cant_rf,
                         "nrmse": nrmse_rf})
```

```python
nrmse_rf.to_csv("../../data/model_results/metrics/rf.csv")
```

## Best

```python
nrmse_rf = pd.read_csv("../../data/model_results/metrics/rf.csv")

best_rf = Model(df, cantons[nrmse_rf["nrmse"].idxmin()])
```

### Plot results

```python
X_combined, y_combined, X_test, y_test, test, pds, epsilon = best_rf.partition()
y_pred_point, lower_bound, upper_bound = best_rf.prediction_intervals(X_combined, y_combined, X_test, epsilon, "rf", pds)
best_rf.serie_temp_canton("rf", lower_bound, upper_bound)
```

### Confidence interval for NRMSE

```python
X_combined, y_combined, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.confint_nrmse("rf", X_combined, y_combined, X_test, y_test, epsilon, n_bootstraps = 500)
```

### Feature importance

```python
X_combined, y_combined, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.var_importance(X_combined, X_test, y_test, "rf", var = "RR")
```

### Feature selection

```python
X_combined, y_combined, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.rfecv_selection("rf", X_combined, y_combined, X_test, y_test, pds, var = "RR")
```
