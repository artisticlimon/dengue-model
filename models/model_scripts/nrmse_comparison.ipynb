# Comparing NRMSE


```python
import pandas as pd
import sys
import numpy as np
from pathlib import Path

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from model_rr import Model
```

```python
cantons = [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115, 116, 117, 118, 119, 120, 201, -202, 203, 204, 205, 206, 207, 208, 209, 210, 211, 212, 213, 214, 215, 216, 301, 302, 303, 304, 305, 306, 307, 308, 401, 402, 403, 404, 405, 406, 407, 408, 409, 410, 501, 502, 503, 504, 505, 506, 507, 508, 509, 510, 511, 601, 602, 603, 604, 605, 606, 607, 608, 609, 610, 611, 701, 702, 703, 704, 705, 706]

cantons = [str(x) for x in cantons]

cantons.append("full")
```

# RF

```python
df = pd.read_csv('../../data/clean/full_data.csv')
```

```python
cod_cant_rf = []
point_rf = []
lower_rf = []
upper_rf = []

for canton in cantons: 
    instancia = Model(df, canton)
    X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = instancia.partition()
    canton_temp, point_temp, lower_temp, upper_temp = instancia.confint_nrmse("rf", X_combined, y_combined, X_test, y_test, epsilon, n_bootstraps = 100)
    cod_cant_rf.append(canton_temp)
    point_rf.append(point_temp)
    lower_rf.append(lower_temp)
    upper_rf.append(upper_temp)
    print(f"Confint ready for {canton}")

confint_rf = pd.DataFrame({"canton": cod_cant_rf,
                         "point": point_rf,
                         "lower": lower_rf, 
                         "upper": upper_rf})
```

```python
confint_rf.to_csv("../../data/model_results/metrics/rf_confint.csv", index = False)
```

# Plots

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def plot_nrmse_confint(model_type):
    confint = pd.read_csv(f"../../data/model_results/metrics/{model_type}_confint.csv")
    confint_nofull = confint[confint["canton"] != "full"].reset_index(drop = True)
    nrmse_nofull = pd.read_csv(f"../../data/model_results/metrics/{model_type}.csv")
    nrmse_nofull["canton"] =  nrmse_nofull["canton"].astype(str)
    confint_full = pd.read_csv(f"../../data/model_results/metrics/{model_type}_confint_full.csv")
    confint_full["canton"] =  confint_full["canton"].astype(str)
    nrmse_full = pd.read_csv(f"../../data/model_results/metrics/{model_type}_full.csv")
    nrmse_full["canton"] =  nrmse_full["canton"].astype(str)
    nrmse_nofull.drop(columns = ["Unnamed: 0"], inplace = True)

    nrmse_confint_nofull = pd.merge(nrmse_nofull, confint_nofull, on = "canton").reset_index(drop = True)
    nrmse_confint_full = pd.merge(nrmse_full, confint_full, on = "canton")

    x = np.arange(len(nrmse_confint_nofull["canton"]))
    width = 0.25

    conf_nofull = np.array(np.abs(nrmse_confint_nofull["nrmse"] - nrmse_confint_nofull["lower"]), np.abs(nrmse_confint_nofull["upper"] - nrmse_confint_nofull["nrmse"]))
    conf_full = np.array(np.abs(nrmse_confint_full["nrmse"] - nrmse_confint_full["lower"]), np.abs(nrmse_confint_full["upper"] - nrmse_confint_full["nrmse"]))

    fig, ax = plt.subplots(figsize=(18, 8))
    ax.errorbar(x, nrmse_confint_nofull["nrmse"], yerr=conf_nofull, fmt='o', color='black', ecolor='gray', elinewidth=2, capsize=5, markersize=8, label = "Model For Each Canton")
    ax.errorbar(x + width, nrmse_confint_full["nrmse"], yerr=conf_full, fmt='o', color='darkred',  ecolor='red', elinewidth=2, capsize=5, markersize=8, label = "Model Trained on Full Data")

    ax.set_xticks(x)
    ax.set_xticklabels(nrmse_confint_nofull["canton"], rotation=90, ha='right', rotation_mode="anchor")
    ax.set_title(f"NRMSE for the {model_type} model by canton")
    ax.set_ylabel('NRMSE')
    ax.set_xlabel('Canton')
    ax.set_ylim([0, 7])
    ax.legend()

    plt.show()
```

## RF

```python
plot_nrmse_confint("rf")
```

## XGB

```python
plot_nrmse_confint("xgb")
```

## HRF

```python
plot_nrmse_confint("hrf")
```

## HXGB

```python
plot_nrmse_confint("hxgb")
```
