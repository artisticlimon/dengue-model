## Save metrics

```python
cod_cant_rf = []
nrmse_rf = []

for canton in cantons: 
    instancia = Model(df, canton)
    X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = instancia.partition()
    cant_temp, nrmse_temp = instancia.calculate_nrmse("rf", epsilon)
    cod_cant_rf.append(canton)
    nrmse_rf.append(nrmse_temp)

nrmse_rf = pd.DataFrame({"canton": cod_cant_rf,
                         "nrmse": nrmse_rf})
```

```python
nrmse_rf.to_csv("../../data/model_results/metrics/rf.csv")
```

## Best

```python
nrmse_rf = pd.read_csv("../../data/model_results/metrics/rf.csv")

nrmse_rf = nrmse_rf[np.isfinite(nrmse_rf)] # Prevent using cantons with an NRMSE undefined because their mean cases is 0
nrmse_rf = nrmse_rf[nrmse_rf["nrmse"].notnull()] # Prevent using cantons with an NRMSE undefined because their mean cases is 0
```

```python
nrmse_rf = pd.read_csv("../../data/model_results/metrics/rf.csv")

best_rf = Model(df, cantons[nrmse_rf["nrmse"].idxmin()])
```

### Plot results

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_rf.partition()
y_pred_point, lower_bound, upper_bound = best_rf.prediction_intervals(X_combined, y_combined, X_test, epsilon, "rf", repeats = 100)
best_rf.serie_temp_canton("rf", y_pred_point, lower_bound, upper_bound)
```

### Confidence interval for NRMSE

```python
print(f"Best canton NRMSE: {round(nrmse_rf.iloc[nrmse_rf["nrmse"].idxmin(), 2], 4)}")
```

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.confint_nrmse("rf", X_combined, y_combined, X_test, y_test, epsilon, n_bootstraps = 100)
```

### Feature importance

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.var_importance(X_test, y_test, "rf", repeats = 100, var = "RR")
```

### Feature selection

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_rf.partition()
best_rf.rfecv_selection("rf", X_combined, y_combined, X_test, y_test, pds, epsilon, repeats = 100, var = "RR")
```

## Save metrics

```python
cod_cant_hxgb = []
nrmse_hxgb = []

for canton in cantons: 
    instancia = Model(df, canton)
    cant_temp, nrmse_temp = instancia.calculate_nrmse("hxgb")
    cod_cant_hxgb.append(canton)
    nrmse_hxgb.append(nrmse_temp)

nrmse_hxgb = pd.DataFrame({"canton": cod_cant_hxgb,
                         "nrmse": nrmse_hxgb})

nrmse_hxgb.to_csv("../../data/model_results/metrics/hxgb.csv")
```

## Best

```python
nrmse_hxgb = pd.read_csv("../../data/model_results/metrics/hxgb.csv")

nrmse_hxgb = nrmse_hxgb[np.isfinite(nrmse_hxgb)] # Prevent using cantons with an NRMSE undefined because their mean cases is 0
nrmse_hxgb = nrmse_hxgb[nrmse_hxgb["nrmse"].notnull()] # Prevent using cantons with an NRMSE undefined because their mean cases is 0
```

```python
best_hxgb = Model(df, cantons[nrmse_hxgb["nrmse"].idxmin()])
```

### Plot results

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_hxgb.partition()
y_pred_point, lower_bound, upper_bound = best_hxgb.prediction_intervals(X_combined, y_combined, X_test, epsilon, "hxgb", repeats = 100)
best_hxgb.serie_temp_canton("hxgb", y_pred_point, lower_bound, upper_bound)
```

### Confidence intervals for NRMSE

```python
print(f"Best canton NRMSE for HXGB: {round(nrmse_hxgb.iloc[nrmse_hxgb["nrmse"].idxmin(), 2], 4)}")
```

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_hxgb.partition()
best_hxgb.confint_nrmse("hxgb", X_combined, y_combined, X_test, y_test, epsilon, n_bootstraps = 100)
```

### Feature importance

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_hxgb.partition()
best_hxgb.var_importance(X_test, y_test, "hxgb", repeats = 100, var = "RR")
```

### Feature selection

```python
X_combined, y_combined, X_train, X_test, y_test, test, pds, epsilon = best_hxgb.partition()
best_hxgb.rfecv_selection("hxgb", X_combined, y_combined, X_test, y_test, pds, repeats = 100, var = "RR")
```
