# Processing the urbanization dataset

```python
import pandas as pd
```

```python
urb = pd.read_excel("../../data/raw/grado_de_urbanidad.xlsx")
```

```python
urb
```

```python
urb = urb.replace({"Urbano": 1, "Rural": 0, "Rural-urbano": 2})
```

```python
urb
```

```python
semanas = [f"2020-{i}" for i in range(1, 54)]
semanas = semanas + [f"2021-{i}" for i in range(1, 54)]
semanas = semanas + [f"2022-{i}" for i in range(1, 54)]
semanas = semanas + [f"2023-{i}" for i in range(1, 54)]
semanas = semanas + [f"2024-{i}" for i in range(1, 54)]
semanas = semanas + [f"2025-{i}" for i in range(1, 54)]
```

```python
semanas
```

```python
for semana in semanas:
    urb[semana] = urb["URBANIZACION"]
urb.drop(columns=["URBANIZACION", "CANTON"], inplace=True)
```

```python
urb
```

```python
urb = pd.melt(urb, id_vars=["ID"], var_name="SEMANA", value_name="URBANIZACION")
```
