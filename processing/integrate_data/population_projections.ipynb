# Weekly Population Projections 2020-2025


```python
import pandas as pd
import numpy as np
```

```python
constants = pd.read_excel("../../data/raw/pob_anual.xlsx", sheet_name="constants")
```

```python
constants.head()
```

```python
semanas_2019 = [f"2019-{i}" for i in range(27, 54)]
semanas = semanas_2019 + [f"2020-{i}" for i in range(1, 54)]
semanas = semanas + [f"2021-{i}" for i in range(1, 54)]
semanas = semanas + [f"2022-{i}" for i in range(1, 54)]
semanas = semanas + [f"2023-{i}" for i in range(1, 54)]
semanas = semanas + [f"2024-{i}" for i in range(1, 54)]
semanas = semanas + [f"2024-{i}" for i in range(1, 54)]
semanas = semanas + [f"2025-{i}" for i in range(1, 54)]
```

```python
for i in range(len(semanas)):
    constants[semanas[i]] = constants["n0"] * np.exp(constants["r"] * (constants["t"] + i / 53 + 2/53))
```

```python
constants.head()
```

```python
pob_prev = constants.drop(columns=["n0", "r", "t"])
pob_prev.drop(columns=semanas_2019, inplace=True)
```

```python
pob_prev.head()
```

```python
pob_prev.head()
```

```python
pob = pd.melt(pob_prev, id_vars="semana", var_name = "canton", value_name="pob")
```

```python
pob["week-canton"] = pob["semana"] + "-" + pob["canton"].astype(str)
pob.drop(pob=["semana", "canton"], inplace=True)
```
