# Join

```python
print(f'Temp max shape: {temp_max.shape}, temp min shape: {temp_min.shape}')
```

```python
temp = pd.merge(temp_max, temp_min, on='week-canton', how='outer')
```

```python
temp.shape
```

```python
temp.head()
```

```python
print(f'Temp shape: {temp.shape}, precip shape: {precip.shape}')
```

```python
temp_precip = pd.merge(temp, precip, on='week-canton', how='right')
```

```python
temp_precip.shape
```

```python
temp_precip.head()
```

```python
print(f'Cases shape: {cases.shape}, rr shape: {rr.shape}, urb shape: {urb.shape}')
```

```python
rr.columns = ['relative_risk', 'week-canton', "classi_rr"]
```

```python
cases_rr = pd.merge(cases, rr, on=['week-canton'], how='outer')
```

```python
cases_rr.shape
```

```python
cases_rr.head()
```

```python
urb.head()
```

```python
urb.columns = ['urb', 'week-canton']
```

```python
demo = pd.merge(cases_rr, urb, on='week-canton', how='right')
```

```python
demo.head()
```

```python
print(f'Demo shape: {demo.shape}, temp_precip shape: {temp_precip.shape}, expected shape: {(temp_precip.shape[0], demo.shape[1] + temp_precip.shape[1] - 1)}')
```

```python
df = pd.merge(demo, temp_precip, on='week-canton', how='right')
```
