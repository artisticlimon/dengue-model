# EDA for number of dengue cases

## Unifying the datasets

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cases2020 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2020", header = 0)
cases2021 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2021", header = 0)
cases2022 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2022", header = 0)
cases2023 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2023", header = 0)
cases2024 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2024", header = 0)
cases2025 = pd.read_excel("../data/raw/dengue_limpio.xlsx", sheet_name="2025", header = 0)
```

```python
cases2020.columns 
```

```python
cases2020.head()
```

```python
cases2020 = cases2020.set_index("ID")
cases2021 = cases2021.set_index("ID")
cases2022 = cases2022.set_index("ID")
cases2023 = cases2023.set_index("ID")
cases2024 = cases2024.set_index("ID")
cases2025 = cases2025.set_index("ID")
```

```python
cases2020_trans = cases2020.transpose()
cases2021_trans = cases2021.transpose()
cases2022_trans = cases2022.transpose()
cases2023_trans = cases2023.transpose()
cases2024_trans = cases2024.transpose()
cases2025_trans = cases2025.transpose()
```

```python
cases2025.tail()
```

```python
fig, ax = plt.subplots(2, 3, figsize=(12, 8))

fig.suptitle("Casos de dengue según semana epidemiológica")

ax[0, 0].plot(cases2020_trans.index[3:], cases2020_trans.iloc[3:, 0])
ax[0, 0].set_title("2020")
ax[0, 0].set_ylabel("Casos")
ax[0, 0].set_xlabel("Semana")

ax[0, 1].plot(cases2021_trans.index[3:], cases2021_trans.iloc[3:, 0])
ax[0, 1].set_title("2021")
ax[0, 1].set_ylabel("Casos")
ax[0, 1].set_xlabel("Semana")

ax[0, 2].plot(cases2022_trans.index[3:], cases2022_trans.iloc[3:, 0])
ax[0, 2].set_title("2022")
ax[0, 2].set_ylabel("Casos")
ax[0, 2].set_xlabel("Semana")

ax[1, 0].plot(cases2023_trans.index[3:], cases2023_trans.iloc[3:, 0])
ax[1, 0].set_title("2023")
ax[1, 0].set_ylabel("Casos")
ax[1, 0].set_xlabel("Semana")

ax[1, 1].plot(cases2024_trans.index[3:], cases2024_trans.iloc[3:, 0])
ax[1, 1].set_title("2024")
ax[1, 2].set_ylabel("Casos")
ax[1, 2].set_xlabel("Semana")

ax[1, 2].plot(cases2025_trans.index[3:], cases2025_trans.iloc[3:, 0])
ax[1, 2].set_title("2025")
ax[1, 2].set_ylabel("Casos")
ax[1, 2].set_xlabel("Semana")

plt.tight_layout()
plt.show()
```

```python
nombres_columnas = ["PROVINCIA Y CANTON", "NUMERO", "TASA"] 
cases2020.columns = nombres_columnas + [f"2020-{i}" for i in range(1, 54)]
cases2021.columns = nombres_columnas + [f"2021-{i}" for i in range(1, 54)]
cases2022.columns = nombres_columnas + [f"2022-{i}" for i in range(1, 54)]
cases2023.columns = nombres_columnas + [f"2023-{i}" for i in range(1, 54)]
cases2024.columns = nombres_columnas + [f"2024-{i}" for i in range(1, 54)]
cases2025.columns = nombres_columnas + [f"2025-{i}" for i in range(1, 54)]
```

```python
cases2020 = cases2020.drop(columns=["PROVINCIA Y CANTON", "NUMERO", "TASA"])
cases2021 = cases2021.drop(columns=["PROVINCIA Y CANTON","NUMERO", "TASA"])              
cases2022 = cases2022.drop(columns=["PROVINCIA Y CANTON","NUMERO", "TASA"])
cases2023 = cases2023.drop(columns=["PROVINCIA Y CANTON","NUMERO", "TASA"])
cases2024 = cases2024.drop(columns=["PROVINCIA Y CANTON","NUMERO", "TASA"])
cases2025 = cases2025.drop(columns=["PROVINCIA Y CANTON","NUMERO", "TASA"])
```

```python
cases2020 = cases2020.reset_index()
cases2021 = cases2021.reset_index()
cases2022 = cases2022.reset_index()
cases2023 = cases2023.reset_index()
cases2024 = cases2024.reset_index()
cases2025 = cases2025.reset_index()
```

```python
casos = pd.merge(cases2020, cases2021, on="ID", how="inner")
casos = pd.merge(casos, cases2022, on="ID", how="inner")
casos = pd.merge(casos, cases2023, on="ID", how="inner")
casos = pd.merge(casos, cases2024, on="ID", how="inner")
casos = pd.merge(casos, cases2025, on="ID", how="inner")
```

```python
casos.head()
```

```python
casos = casos.set_index("ID")
```

```python
casos.head()
```

```python
casos_trans = casos.transpose()
```

```python
casos = casos_trans.reset_index()
```

```python
casos = casos.rename(columns={"index": "semana"})
```

```python
casos = casos[~casos["semana"].isin(["2020-53", "2021-53", "2022-53", "2023-53", "2024-53", "2025-53"])]
```

```python
fig, ax = plt.subplots(4, 2, figsize=(20, 24))

fig.suptitle("Distribución de los casos de dengue por cantón")

ax[0, 0].boxplot(casos.iloc[:, 3:23], tick_labels = casos.columns[3:23])
ax[0, 0].set_title("Distribución de los casos de los cantones de San José")

ax[0, 1].boxplot(casos.iloc[:, 24:40], tick_labels = casos.columns[24:40])
ax[0, 1].set_title("Distribución de los casos de los cantones de Alajuela")

ax[1, 0].boxplot(casos.iloc[:, 41:49], tick_labels = casos.columns[41:49])
ax[1, 0].set_title("Distribución de los casos de los cantones de Cartago")

ax[1, 1].boxplot(casos.iloc[:, 50:60], tick_labels = casos.columns[50:60])
ax[1, 1].set_title("Distribución de los casos de los cantones de Heredia")

ax[2, 0].boxplot(casos.iloc[:, 61:72], tick_labels = casos.columns[61:72])
ax[2, 0].set_title("Distribución de los casos de los cantones de Guanacaste")

ax[2, 1].boxplot(casos.iloc[:, 73:84], tick_labels = casos.columns[73:84])
ax[2, 1].set_title("Distribución de los casos de los cantones de Puntarenas")

ax[3, 0].boxplot(casos.iloc[:, 85:91], tick_labels = casos.columns[85:91])
ax[3, 0].set_title("Distribución de los casos de los cantones de Limón")

plt.show()
```
