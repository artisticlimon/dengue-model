## <a id='toc1_1_'></a>[Load datasets](#toc0_)

```python
# REGULAR
cases = pd.read_csv('../data/interim/casos_dengue.csv')

rr = pd.read_csv('../data/interim/relative_risk.csv')
rr.drop(columns=["Unnamed: 0"], inplace=True)
rr = rr.rename(columns = {"week-canton": "week_canton"})

ndvi = pd.read_csv('../data/interim/NDVI_int.csv')
ndvi.drop(columns=["Unnamed: 0"], inplace=True)

tmax = pd.read_csv('../data/interim/tmax_weekly.csv')
tmax.drop(columns=["Unnamed: 0"], inplace=True)

tmin = pd.read_csv('../data/interim/tmin_weekly.csv')
tmin.drop(columns=["Unnamed: 0"], inplace=True)

temp_prom = pd.read_csv('../data/interim/temp_prom.csv')
temp_prom.drop(columns=["Unnamed: 0"], inplace=True)

sst = pd.read_csv('../data/interim/SST_int.csv')
sst.drop(columns=["Unnamed: 0"], inplace=True)

keywords = pd.read_csv('../data/interim/trends_keywords.csv')
keywords.drop(columns=["Unnamed: 0"], inplace=True)

precip = pd.read_csv('../data/interim/precip_weekly.csv')
precip.drop(columns=["Unnamed: 0"], inplace=True)

urb = pd.read_csv('../data/interim/urbanization.csv')
```

```python
# LAGS
cases_lags = pd.read_csv('../data/interim/lags/casos_dengue_lags.csv')
cases_lags.drop(columns=["Unnamed: 0"], inplace=True)

rr_lags = pd.read_csv('../data/interim/lags/relative_risk_lags.csv')
rr_lags.drop(columns=["Unnamed: 0"], inplace=True)

ndvi_lags = pd.read_csv('../data/interim/lags/NDVI_lags.csv')
ndvi_lags.drop(columns=["Unnamed: 0"], inplace=True)

tmax_lags = pd.read_csv('../data/interim/lags/tmax_lags.csv')
tmax_lags.drop(columns=["Unnamed: 0"], inplace=True)

tmin_lags = pd.read_csv('../data/interim/lags/tmin_lags.csv')
tmin_lags.drop(columns=["Unnamed: 0"], inplace=True)

temp_prom_lags = pd.read_csv('../data/interim/lags/temp_prom_lags.csv')
temp_prom_lags.drop(columns=["Unnamed: 0"], inplace=True)

sst_lags = pd.read_csv('../data/interim/lags/sst_lags.csv')
sst_lags.drop(columns=["Unnamed: 0"], inplace=True)

precip_lags = pd.read_csv('../data/interim/lags/precip_lags.csv')
precip_lags.drop(columns=["Unnamed: 0"], inplace=True)

keywords_lags = pd.read_csv('../data/interim/lags/trends_keywords_lags.csv')
keywords_lags.drop(columns=["Unnamed: 0"], inplace=True)
```

# <a id='toc3_'></a>[Join](#toc0_)

```python
print(f'Temp max shape: {tmax.shape}, temp min shape: {tmin.shape}')
```

```python
temp = pd.merge(tmin, tmax, on='week_canton', how='inner')
temp = pd.merge(temp, temp_prom, on='week_canton', how='inner')
```

```python
demo = pd.merge(cases_rr, urb, on='week_canton', how='inner')
demo = pd.merge(demo, keywords, on='week_canton', how='inner')
```

```python
demo.shape
```

```python
print(f'Demo shape: {demo.shape}, temp_precip shape: {temp_precip.shape}, expected shape: {(temp_precip.shape[0], demo.shape[1] + temp_precip.shape[1] - 1)}')
```

```python
demo_temp_precip = pd.merge(demo, temp_precip, on='week_canton', how='inner')
```

```python
demo_temp_precip.shape
```

```python
sst_ndvi = pd.merge(sst, ndvi, on='week_canton', how='inner')
```

```python
sst_ndvi.shape
```

```python
df_no_lag = pd.merge(demo_temp_precip, sst_ndvi, on='week_canton', how='inner')
```

```python
df_no_lag.shape
```

```python
cases_lags.shape
```

```python
df_lag1 = pd.merge(df_no_lag, cases_lags, on='week_canton', how='inner')
```

```python
df_lag1.shape
```

```python
df_lag2 = pd.merge(df_lag1, rr_lags, on='week_canton', how='inner')
df_lag3 = pd.merge(df_lag2, ndvi_lags, on='week_canton', how='inner')
df_lag4 = pd.merge(df_lag3, tmax_lags, on='week_canton', how='inner')
df_lag5 = pd.merge(df_lag4, tmin_lags, on='week_canton', how='inner')
df_lag6 = pd.merge(df_lag5, sst_lags, on='week_canton', how='inner')
df_lag7 = pd.merge(df_lag6, temp_prom_lags, on='week_canton', how='inner')
df_lag8 = pd.merge(df_lag7, keywords_lags, on='week_canton', how='inner')
df = pd.merge(df_lag8, precip_lags, on='week_canton', how='inner')
```

```python
df['urb'] = df['urb'].astype('category')
df['clasi_rr'] = df['clasi_rr'].astype('category')
df['clasi_rr_lag_1'] = df['clasi_rr_lag_1'].astype('category')
df['clasi_rr_lag_2'] = df['clasi_rr_lag_2'].astype('category')
df['clasi_rr_lag_3'] = df['clasi_rr_lag_3'].astype('category')
df['clasi_rr_lag_4'] = df['clasi_rr_lag_4'].astype('category')
df['clasi_rr_lag_5'] = df['clasi_rr_lag_5'].astype('category')
df['clasi_rr_lag_6'] = df['clasi_rr_lag_6'].astype('category')
df['clasi_rr_lag_7'] = df['clasi_rr_lag_7'].astype('category')
df['clasi_rr_lag_8'] = df['clasi_rr_lag_8'].astype('category')
```
